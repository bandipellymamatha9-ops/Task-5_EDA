# Exploratory Data Analysis (EDA) – Titanic-Style Passenger Sample

**Objective:** Explore data using statistics and visualisations.

**Dataset note:** This is a synthetic Titanic-style teaching dataset with 500 sample rows. It is not the original historical Titanic passenger dataset. A few missing values are included for data-cleaning practice.

## 1. Import libraries and load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("titanic_passengers_sample.csv")
df.head()
```

## 2. Basic dataset exploration

```python
print("Dataset shape:", df.shape)
df.info()
display(df.describe())
display(df["survived"].value_counts())
```

## 3. Missing values and simple cleaning

```python
print("Missing values before cleaning:")
display(df.isnull().sum())

df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

print("Missing values after cleaning:")
display(df.isnull().sum())
```

## 4. Univariate analysis

```python
plt.figure(figsize=(6, 4))
df["survived"].value_counts().sort_index().plot(kind="bar")
plt.title("Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Passenger count")
plt.show()
```

## 5. Bivariate analysis

```python
survival_by_sex = df.groupby("sex")["survived"].mean()
survival_by_sex.plot(kind="bar", title="Survival Rate by Sex")
plt.ylabel("Survival rate")
plt.show()

survival_by_class = df.groupby("pclass")["survived"].mean()
survival_by_class.plot(kind="bar", title="Survival Rate by Passenger Class")
plt.ylabel("Survival rate")
plt.show()
```

## 6. Boxplot and scatterplot

```python
sns.boxplot(data=df, x="pclass", y="fare")
plt.title("Fare Distribution by Passenger Class")
plt.show()

plt.figure(figsize=(6, 4))
plt.scatter(df["age"], df["fare"], alpha=0.6)
plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

## 7. Correlation heatmap and pairplot

```python
numeric_data = df[["survived", "pclass", "age", "sibsp", "parch", "fare"]].copy()
sns.heatmap(numeric_data.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()

sns.pairplot(df[["age", "fare", "pclass", "survived"]], hue="survived")
plt.show()
```

## 8. Observations and conclusion

- The dataset contains 500 passenger sample records.
- The survival column has two categories: 0 (did not survive) and 1 (survived).
- Survival rates differ between female and male passengers in this sample.
- Survival rates also differ across passenger classes.
- Age and embarkation columns contain missing values; median and mode are used to fill them.
- The fare boxplot helps identify spread and unusually high values.
- The heatmap gives a quick view of numerical relationships. Correlation does not prove cause and effect.

**Conclusion:** EDA helps us understand the structure, quality, distributions and relationships in data before further analysis. These results describe this synthetic sample only.
